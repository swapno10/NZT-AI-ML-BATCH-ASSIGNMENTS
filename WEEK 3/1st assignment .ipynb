```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import string
```

```python
#input:-

texts="""Language is powerful: it connects people, expresses ideas, and preserves culture. When we speak or write, we don’t just use words—we rely on punctuation marks, too! Commas separate ideas, periods end statements, and question marks ask things like, “What do you mean?” Sometimes writers add emphasis with dashes—like this—or extra information (in parentheses). Apostrophes show possession or contractions (for example, it’s, don’t, and Minakshi’s book). Without punctuation, sentences could become confusing; however, with proper marks, writing becomes clearer, smoother, and more engaging. Imagine reading a story without punctuation: strange, right? The characters might shout, “Run!” or whisper, “Be quiet...” while the narrator explains what’s happening. Lists appear often—apples, oranges, bananas, and grapes—making ideas organized. Writers may also compare ideas; for instance, short sentences create tension, whereas longer ones add detail. Sometimes they add a surprising twist—unexpected, dramatic, unforgettable! In the end, punctuation isn’t just decoration; it’s a system that guides readers, shapes meaning, and makes communication effective."""
```

```python
# 1.removing punctuation marks

text=text.translate(str.maketrans('', '', string.punctuation))
print(text)
```

```python
# 2.stop word removal

stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
tokens=word_tokenize(text)
stop_words=set(stop_words)
filter=[w for w in tokens if w.lower() not in stop_words]

print(filter)
```

```python
# 3.stemming

stemmer=PorterStemmer()
stemmed_words=[stemmer.stem(word) for word in filter]

print(stemmed_words)
```

```python
# 4.lemmatization

lemmatizer=WordNetLemmatizer()
lemmatized_words=[lemmatizer.lemmatize(word,pos='v') for word in filter]

print(lemmatized_words)
```

```python
from nltk import pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

lemmatizer1=WordNetLemmatizer()

def get_pos(word):
    tag=pos_tag([word])[0][1][0].upper()

    tag_dict={
        "J": wordnet.ADJ,
        "N": wordnet.NOUN,
        "V": wordnet.VERB,
        "R": wordnet.ADV
    }

    return tag_dict.get(tag,wordnet.NOUN)

lemmatized_words1=[lemmatizer1.lemmatize(word,get_pos(word)) for word in filter]

print(lemmatized_words)
```

```python
#Word2Vec


from gensim.models import Word2Vec
model= Word2Vec(
    [words],
    vector_size=100,
    window=5,
    min_count=1,
    workers=500
)
```
